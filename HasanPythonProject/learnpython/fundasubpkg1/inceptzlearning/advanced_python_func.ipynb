```python
# Local Scope Example
a = 10 # Global variable
def fun1():
    a = 100 # Local variable overrides inside
    print("Inside:", a)

fun1() # Prints 100
print("Outside:", a) # Prints 10
```

```python
# Global Keyword Example
a = 10
def fun2():
    global a # Reference the global variable
    a = 100 # Modifies global variable directly
    print("Inside:", a)

fun2() # Prints 100
print("Outside:", a) # Prints 100 (Modified!)
```

6. Higher Order Functions (HOF)

Type 1: Function as an Argument

Instead of just passing static values, you pass an entire function as an argument. This enables high reusability.

```python
def percent_calc(pct):
    return pct / 100
# calc_func is a function passed as argument
def calc_gross(sal, pct, calc_func):
    fraction = calc_func(pct)
    return sal + (sal * fraction)
print(calc_gross(100000, 10, percent_calc))
```

Type 2: Returning a Function (Currying)

A function that constructs and returns another function. Useful for creating configurable partial functions.

```python
def calculator(typ):
    if typ == "add":
        def add_func(a, b):
            return a + b
        return add_func # Returns function itself

# partial_func becomes the add_func logic
partial_func = calculator("add")
print("Result:", partial_func(10, 20))
```

```python
# lambda args: expression
calc_bonus = lambda sal, pct: sal + (sal * (pct/100))
# Quick one-liner invocations
print(calc_bonus(10000, 10))
print(calc_bonus(20000, 10))
```

```python
#HOF Type1: A function that take another function as an input argument
#basically a function can be passed as input to another function
def fun1():
    return 'inceptez'

def fun2(arg):#arg=fun1 - arg is going to assigned with fun1 code
    print(arg)

fun2(fun1)#arg=fun1 code - call by reference
fun2(fun1())#arg='inceptez' - call by value (Composable)
```

```python
#HOF Type2: A function that can return another function as an output
def parent():
    print("parent is running with some logic & returning child")
    def child(a):
        print(f"child is running now with a param of {a}")
    return child
childvar=parent()#run parent & returns child
childvar(100)
```

```python
#HOF Type1 : A Function take another function as an argument
def bonus_calc(sal):
    return sal*.10
def inc_calc():
    return 3000
def total_sal(sal,fun,fun1):
    #bonus_inc_sal=fun(sal)#HOF type1
    final_sal=sal+fun(sal)+fun1()
    return final_sal
print(total_sal(10000,bonus_calc,inc_calc))
```

```python
#HOF Type2: A function that can return another function as an output
#Reallife Example: GPay payment -> 2 levels of protection - level1: entering into gpay app, level2: entering pin
actual_fingerprint='xyz'#closure
def entering_gpay(input_fingerprint):
    print("Analyzing the fingerprint")
    if input_fingerprint==actual_fingerprint:#Closure affects the function result
        print(f"{input_fingerprint} fingerprint matched, please proceed.")
        print("I will return you the payment function enter_pin()")
        actual_pin='1234'
        def enter_pin_payment(pin,payment):
            if actual_pin==pin:
                print(f"Pin entered {pin} matches , hence allowing you to make payment {payment}")
        return enter_pin_payment
    else:
        print(f"{input_fingerprint} fingerprint not matched")


payment_proc = entering_gpay('xyz')
payment_proc('1234',10000)
```

```python
#Nested Function
def briyani():
    print("returning plain briyani")
    def salt():
        print("adding salt")
        def chilli():
            print("adding chilli")
            return 'ready to consume briyani'
        return chilli
    return salt

plainbriyani=briyani()()()
print(plainbriyani)
```
