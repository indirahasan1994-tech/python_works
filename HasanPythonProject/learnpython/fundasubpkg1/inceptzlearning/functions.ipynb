```python
def swiggy_bill(cart_value: int):
    min_cart_amt= 600
    disc_pct_whole =10
    max_disc_amt=100
    disc_pct = disc_pct_whole/100
    #cart_value = int(input("The amount for the selected product is:"))
    disc_amt = cart_value*disc_pct
    if cart_value < min_cart_amt:
        out = "You cart value is not eligible for an offer, please add products worth"+str(min_cart_amt-cart_value)+" Rupee more to avail this offer"
        return out
    else:
        if disc_amt <= max_disc_amt:
            final_price = cart_value - disc_amt
            out ="Final cart value after applying 10% discount is"+" "+str(disc_amt)+" is:"+ str(final_price)+", please proceed for payment"
            return out
            #print(f"Final cart value after applying 10% discount is ({disc_amt}) is: {final_price}, please proceed for payment")
        else:
            final_price = cart_value - max_disc_amt
            out ="Final cart value after applying max discount amt"+" "+str(max_disc_amt)+" is:"+ str(final_price)+", please proceed for payment"
            return out
            #print(f"Final cart value after applying max discount amt({max_disc_amt}) is: {final_price}, please proceed for payment")
```
